python list ko hum arry be bolta hain 
numy array is tarha use hota hain { pn_array = n.array} np hum ne import kia hai

```python
py_list = [1,2,3,4]
print("python list multiplication" , py_list * 2)

np_array  = np.array ([2,4,5,6])  # har element 2 say ye jo number mila ga os say multipy ho ga
print("python arry multiplication" , np_array * 2)


start = time.time()
py_list = [i*2 for i in  range(1000000) ]
print("\n List operation time :" , time.time()- start) # ye hai hum basic classic list ko kesa * kr sakta hain

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operation time :" , time.time()- start)

```

Creating arrays from scratch

```python
zeros = np.zeros((7,8)) # is me ye hai sary zero show ho ga like 7 rows 0 ki 8 column 0 ka
print("Zeros array : \n" , zeros)

ones = np.ones((2,4))
print("Ones array : \n " , ones)

full = np.full((3,4) ,7) # 7 ye wo value hai jo ap ne 3,4 me rakhi hai means ap is ko change be kr sakta ho
print("Full array : \n " , full)

random = np.random.random((1,6))
print("Random array : \n " , random)

sequence = np.arange (0,12,4)
print("sequence array : \n " , sequence)

```

Vector , Matrix & Tensor

```python
vector = np.array([1,2,3]) # (1D array)
print("Vector \n" , vector)

matrix = np.array([[1,2,3],# (2D array)
                    [4,5,6]])
print("Matrix \n" , matrix)

tensor = np.array([[[1,2,3],[4,5,6],# 3D ya higher dimensional array
                    [1,2,3],[3,2,1]]])
print("Tensor \n" , tensor)
```

Array properties

```python
array = np.array([[1,2,3],
                  [3,22,1]])
print("Shape :" ,array.shape)
print("Dementation :" ,array.ndim)
print("Size:" ,array.size)
print("Data_type", array.dtype)
```

Array Reshaping

```python
array = np.arange(12)
print("Orignal Array", array )

reshaped = array.reshape((4,3))
print("Reshaped Array" , reshaped)

flattened = reshaped.flatten()
print("Flattend" , flattened) # ye copy array hota hai
 
raveled = reshaped.ravel()
print("Raveled" , raveled) # ye orignal 

#Tranpose
transpose = reshaped.T
print("Transpose" ,transpose)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000],  # Chai Point
])

print("***** Yamin`s Food Business Analysis*****")
print("\n sale shape" , sales_data.shape)
print("\n sale Datta" , sales_data[1:4])
```

```python
# total sales per year
print(np.sum(sales_data, axis = 1)) # agar me axis 0 kru to column vice add ho ga agar 1 kru to row
year_total = np.sum(sales_data[:, 1:] , axis =0)
# ye dono same kam he krta hain
print(year_total)
```

```python
min_sale = np.min(sales_data[: , 1:] , axis = 1)
print(min_sale)


max_sale = np.max(sales_data[: , 2:] , axis =1)
print(max_sale)
```

```python
# is me sales add hoti hain like 1 saal ki sale me 2nd ki add ki new 
# value mila same next next chalta raha ga
cumsum = np.cumsum(sales_data[: , 1:] , axis = 1)
print(cumsum) 
```

```python
plt.Figure(figsize=(10,5))
plt.plot(np.mean(cumsum , axis = 0))
plt.title("Average cumulative sale across all restrunt")
plt.xlabel("Years")
plt.grid(True)
plt.show()
```
